```python
# 8. Employee Emails
```

```python
# 12. Student Attendance
```

```python
attendance = {
    "John": True,
    "Alice": False,
    "Bob": True,
    "David": None
}
```

```python
# Questions

Check whether Bob's attendance is available.
Find students whose attendance has not been recorded.
Count present students.
Count absent students.
```

```python
is_bob_available = attendance.get("Bob") is not None
print(f"Is Bob's attendance available? {is_bob_available}")

unrecorded_students = [student for student, status in attendance.items() if status is None]
print(f"Students with unrecorded attendance: {unrecorded_students}")

present_count = sum(1 for status in attendance.values() if status is True)
print(f"Count of present students: {present_count}")

absent_count = sum(1 for status in attendance.values() if status is False)
print(f"Count of absent students: {absent_count}")
```

```python
# 13. Movie Ratings
```

```python
ratings = {
    "Movie A": 4.5,
    "Movie B": None,
    "Movie C": 3.8,
    "Movie D": None
}
```

```python
# Questions

Check whether Movie C has a rating.
Find all movies with no ratings.
Count unrated movies.
Replace all missing ratings with 0.
```

```python
print("Movie C has rating:", ratings.get("Movie C") is not None)

unrated_movies = [movie for movie, rating in ratings.items() if rating is None]
print("Movies with no ratings:", unrated_movies)

print("Count of unrated movies:", len(unrated_movies))

for movie in ratings:
    if ratings[movie] is None:
        ratings[movie] = 0

print("Updated ratings:", ratings)
```

```python
Q1.Write a program to check whether Employee ID 1050 exists.
Q2.Ask the user to enter an Employee ID and display whether it exists.
Q3.If the Employee ID exists, display all the employee details.
Q4.If the Employee ID does not exist, display:Employee not found.
Q5.Display the salary of Employee 1032.
Q6.Display the department of Employee 1087.
Q7.Print only the email address of Employee 1015.
Q8.Find all employees whose email address is missing (None).
Q9.Find all employees whose phone number is missing.
Q10.Count how many employees have incomplete records (missing email or phone).
Q11.Replace all missing email addresses with: Not Available
Q12.Find all employees working in the IT department.
Q13.Find all employees working in the Finance department.
Q14.Display the names of all employees managed by Manager_4.
Q15.Find all employees older than 50 years.
Q16.Find all employees younger than 30 years.
Q17.Find the employee with the highest salary.
Q18.Find the employee with the lowest salary.
Q19.Calculate the average salary of all employees.
Q20.Display employees earning more than ₹100,000.
Q21.Display employees earning less than ₹50,000.
Q22.Count how many employees belong to each department.

Example output

HR : 18
Finance : 15
IT : 22
Sales : 17
Marketing : 14
Operations : 14

Q23.Count how many employees are managed by each manager.
Q24. Find which department has the largest number of employees.
Q25.Increase the salary of every employee by 10%.
Q26.Update the department of Employee 1040 to Research.
Q27.Add a new employee to the dictionary.
Q28.Delete Employee 1018 from the dictionary.
Q29.Update the phone number of Employee 1075.
```

```python
from collections import Counter
# Q1
print("1050" in employees)

# Q2
emp_id = input("Enter Employee ID: ")
print(emp_id in employees)

# Q3 & Q4
emp_id = input("Enter Employee ID: ")
if emp_id in employees:
    print(employees[emp_id])
else:
    print("Employee not found.")

# Q5
print(employees.get("1032", {}).get("Salary"))

# Q6
print(employees.get("1087", {}).get("Department"))

# Q7
print(employees.get("1015", {}).get("Email"))

# Q8
print([k for k, v in employees.items() if v["Email"] is None])

# Q9
print([k for k, v in employees.items() if v["Phone"] is None])

# Q10
print(sum(1 for v in employees.values() if v["Email"] is None or v["Phone"] is None))

# Q11
for emp in employees.values():
    if emp["Email"] is None:
        emp["Email"] = "Not Available"

# Q12
print([k for k, v in employees.items() if v["Department"] == "IT"])

# Q13
print([k for k, v in employees.items() if v["Department"] == "Finance"])

# Q14
print([v["Name"] for v in employees.values() if v["Manager"] == "Manager_4"])

# Q15
print([k for k, v in employees.items() if v["Age"] > 50])

# Q16
print([k for k, v in employees.items() if v["Age"] < 30])

# Q17
print(max(employees.items(), key=lambda x: x[1]["Salary"]))

# Q18
print(min(employees.items(), key=lambda x: x[1]["Salary"]))

# Q19
salaries = [v["Salary"] for v in employees.values()]
print(sum(salaries) / len(salaries))

# Q20
print([k for k, v in employees.items() if v["Salary"] > 100000])

# Q21
print([k for k, v in employees.items() if v["Salary"] < 50000])

# Q22
dept_counts = Counter(v["Department"] for v in employees.values())
for dept, count in dept_counts.items():
    print(f"{dept} : {count}")

# Q23
mgr_counts = Counter(v["Manager"] for v in employees.values())
for mgr, count in mgr_counts.items():
    print(f"{mgr} : {count}")

# Q24
print(dept_counts.most_common(1)[0][0])

# Q25
for emp in employees.values():
    emp["Salary"] = round(emp["Salary"] * 1.10, 2)

# Q26
if "1040" in employees:
    employees["1040"]["Department"] = "Research"

# Q27
employees["1101"] = {
    "Name": "Employee_1101",
    "Department": "IT",
    "Age": 28,
    "Salary": 85000,
    "Email": "employee1101@company.com",
    "Phone": "9899001122",
    "Manager": "Manager_1"
}

# Q28
employees.pop("1018", None)

# Q29
if "1075" in employees:
    employees["1075"]["Phone"] = "9988776655"
```

```python
# Generating a report  TRY THIS OUT, else we will see it in class

Q30.Generate a report (HINT - USE -> 'print') showing:

    Total Employees
    Employees with Missing Email
    Employees with Missing Phone
    Average Salary
    Highest Salary
    Lowest Salary

Q31. Generate a report listing all employees department-wise.

Example

IT
-----
Employee_1001
Employee_1005
Employee_1022

Finance
---------
Employee_1008
Employee_1012
...

    
Q32. Generate a report showing employees sorted by salary (highest to lowest).
Q33. Find all employees whose names contain the word "25".
Q34. Find all employees whose salary lies between ₹60,000 and ₹80,000.
Q35. Find managers supervising more than 8 employees.
Q36. Find the department with the highest average salary.
Q37. Identify employees with both email and phone number missing.
Q38. Find duplicate phone numbers (if any).
Q39. Find employees whose email domain is company.com.
Q40. Create a new dictionary containing only employees from the IT department.
```

```python
from collections import defaultdict, Counter

# Q30
salaries = [v["Salary"] for v in employees.values()]
total_emp = len(employees)
missing_email = sum(1 for v in employees.values() if v["Email"] is None or v["Email"] == "Not Available")
missing_phone = sum(1 for v in employees.values() if v["Phone"] is None)
avg_salary = sum(salaries) / total_emp if total_emp else 0
highest_salary = max(salaries) if salaries else 0
lowest_salary = min(salaries) if salaries else 0

print("--- EMPLOYEE REPORT ---")
print("Total Employees:", total_emp)
print("Employees with Missing Email:", missing_email)
print("Employees with Missing Phone:", missing_phone)
print("Average Salary:", round(avg_salary, 2))
print("Highest Salary:", highest_salary)
print("Lowest Salary:", lowest_salary)

# Q31
dept_map = defaultdict(list)
for emp in employees.values():
    dept_map[emp["Department"]].append(emp["Name"])

for dept, emp_list in dept_map.items():
    print(f"\n{dept}")
    print("-" * len(dept))
    for name in emp_list:
        print(name)

# Q32
sorted_by_salary = sorted(employees.values(), key=lambda x: x["Salary"], reverse=True)
for emp in sorted_by_salary:
    print(f"{emp['Name']}: ₹{emp['Salary']}")

# Q33
print([v["Name"] for v in employees.values() if "25" in v["Name"]])

# Q34
print([v["Name"] for v in employees.values() if 60000 <= v["Salary"] <= 80000])

# Q35
mgr_counts = Counter(v["Manager"] for v in employees.values())
print([mgr for mgr, count in mgr_counts.items() if count > 8])

# Q36
dept_salaries = defaultdict(list)
for emp in employees.values():
    dept_salaries[emp["Department"]].append(emp["Salary"])

dept_avg_salaries = {dept: sum(sals) / len(sals) for dept, sals in dept_salaries.items()}
highest_avg_dept = max(dept_avg_salaries.items(), key=lambda x: x[1])[0]
print(highest_avg_dept)

# Q37
print([v["Name"] for v in employees.values() if (v["Email"] is None or v["Email"] == "Not Available") and v["Phone"] is None])

# Q38
phones = [v["Phone"] for v in employees.values() if v["Phone"] is not None]
phone_counts = Counter(phones)
print([phone for phone, count in phone_counts.items() if count > 1])

# Q39
print([v["Name"] for v in employees.values() if v["Email"] and str(v["Email"]).endswith("company.com")])
```

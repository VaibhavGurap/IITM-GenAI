```python
# 1. Print the location of GTU_India.
print(university_system["universities"]["GTU_India"]["location"])

# 2. Print the dean of the engineering faculty in GTU_Singapore.
print(university_system["universities"]["GTU_Singapore"]["faculties"]["engineering"]["dean"])

# 3. Print the HOD of the computer_science department.
print(university_system["universities"]["GTU_India"]["faculties"]["engineering"]["departments"]["computer_science"]["hod"])

# 4. Print all labs under computer_science.
labs = university_system["universities"]["GTU_India"]["faculties"]["engineering"]["departments"]["computer_science"]["labs"]
print(list(labs.keys()))

# 5. Print the funding amount of the "DeepVision" project.
print(university_system["universities"]["GTU_India"]["faculties"]["engineering"]["departments"]["computer_science"]["labs"]["AI Lab"]["research_projects"]["DeepVision"]["funding"])

# 6. Print all research project names in GTU_India.
gtu_india_projects = []
for faculty in university_system["universities"]["GTU_India"]["faculties"].values():
    for dept in faculty.get("departments", {}).values():
        for lab in dept.get("labs", {}).values():
            gtu_india_projects.extend(lab.get("research_projects", {}).keys())
print(gtu_india_projects)

# 7. Print all ongoing projects in GTU_India.
ongoing_india = []
for faculty in university_system["universities"]["GTU_India"]["faculties"].values():
    for dept in faculty.get("departments", {}).values():
        for lab in dept.get("labs", {}).values():
            for proj_name, proj_info in lab.get("research_projects", {}).items():
                if proj_info.get("status") == "ongoing":
                    ongoing_india.append(proj_name)
print(ongoing_india)

# 8. Calculate the total funding for all projects in AI Lab.
ai_lab_projects = university_system["universities"]["GTU_India"]["faculties"]["engineering"]["departments"]["computer_science"]["labs"]["AI Lab"]["research_projects"]
total_ai_funding = sum(proj["funding"] for proj in ai_lab_projects.values())
print(total_ai_funding)

# 9. Print all researchers working under Systems Lab.
systems_lab_projects = university_system["universities"]["GTU_India"]["faculties"]["engineering"]["departments"]["computer_science"]["labs"]["Systems Lab"]["research_projects"]
systems_researchers = []
for proj in systems_lab_projects.values():
    systems_researchers.extend(proj["researchers"])
print(systems_researchers)

# 10. Count total students enrolled in all computer_science courses.
cs_courses = university_system["universities"]["GTU_India"]["faculties"]["engineering"]["departments"]["computer_science"]["courses"]
total_cs_students = sum(course["students_enrolled"] for course in cs_courses.values())
print(total_cs_students)

# 11. Print all university names in the system.
print(list(university_system["universities"].keys()))

# 12. Print all faculty names under GTU_India.
print(list(university_system["universities"]["GTU_India"]["faculties"].keys()))

# 13. Print all department names across all universities.
departments = []
for uni in university_system["universities"].values():
    for faculty in uni.get("faculties", {}).values():
        departments.extend(faculty.get("departments", {}).keys())
print(list(set(departments)))

# 14. Print all deans in the entire system.
deans = []
for uni in university_system["universities"].values():
    for faculty in uni.get("faculties", {}).values():
        if "dean" in faculty:
            deans.append(faculty["dean"])
print(deans)

# 15. Print all project names across both universities.
all_projects = []
for uni in university_system["universities"].values():
    for faculty in uni.get("faculties", {}).values():
        for dept in faculty.get("departments", {}).values():
            for lab in dept.get("labs", {}).values():
                all_projects.extend(lab.get("research_projects", {}).keys())
print(all_projects)

# 16. Add a new research project under AI Lab.
university_system["universities"]["GTU_India"]["faculties"]["engineering"]["departments"]["computer_science"]["labs"]["AI Lab"]["research_projects"]["NeuralNet"] = {
    "funding": 200000,
    "researchers": ["Aman", "Priya"],
    "status": "ongoing"
}

# 17. Change the status of "DeepVision" to completed.
university_system["universities"]["GTU_India"]["faculties"]["engineering"]["departments"]["computer_science"]["labs"]["AI Lab"]["research_projects"]["DeepVision"]["status"] = "completed"

# 18. Increase funding of "CloudScale" by 50,000.
university_system["universities"]["GTU_India"]["faculties"]["engineering"]["departments"]["computer_science"]["labs"]["Systems Lab"]["research_projects"]["CloudScale"]["funding"] += 50000

# 19. Add a new university to the system.
university_system["universities"]["GTU_UK"] = {
    "location": "London",
    "ranking": 10,
    "faculties": {}
}

# 20. Add a new course under finance department.
university_system["universities"]["GTU_India"]["faculties"]["management"]["departments"]["finance"]["courses"]["Investment Analysis"] = {
    "credits": 4,
    "students_enrolled": 60
}

# 21. Find the project with the highest funding.
highest_funding_proj = None
max_funding = -1

for uni in university_system["universities"].values():
    for faculty in uni.get("faculties", {}).values():
        for dept in faculty.get("departments", {}).values():
            for lab in dept.get("labs", {}).values():
                for proj_name, proj_info in lab.get("research_projects", {}).items():
                    if proj_info["funding"] > max_funding:
                        max_funding = proj_info["funding"]
                        highest_funding_proj = proj_name

print(highest_funding_proj, max_funding)

# 22. Find total funding across all universities.
total_system_funding = 0
for uni in university_system["universities"].values():
    for faculty in uni.get("faculties", {}).values():
        for dept in faculty.get("departments", {}).values():
            for lab in dept.get("labs", {}).values():
                for proj_info in lab.get("research_projects", {}).values():
                    total_system_funding += proj_info["funding"]

print(total_system_funding)

# 23. Find the university with the best ranking.
best_uni = min(university_system["universities"].items(), key=lambda x: x[1]["ranking"])[0]
print(best_uni)

# 24. Print all completed projects.
completed_projects = []
for uni in university_system["universities"].values():
    for faculty in uni.get("faculties", {}).values():
        for dept in faculty.get("departments", {}).values():
            for lab in dept.get("labs", {}).values():
                for proj_name, proj_info in lab.get("research_projects", {}).items():
                    if proj_info.get("status") == "completed":
                        completed_projects.append(proj_name)

print(completed_projects)

# 25. Count total number of researchers in GTU_India.
india_researchers = set()
for faculty in university_system["universities"]["GTU_India"]["faculties"].values():
    for dept in faculty.get("departments", {}).values():
        for lab in dept.get("labs", {}).values():
            for proj_info in lab.get("research_projects", {}).values():
                india_researchers.update(proj_info.get("researchers", []))

print(len(india_researchers))

# 26. Find which lab has the maximum funding.
lab_fundings = {}
for uni in university_system["universities"].values():
    for faculty in uni.get("faculties", {}).values():
        for dept in faculty.get("departments", {}).values():
            for lab_name, lab_info in dept.get("labs", {}).items():
                funding_sum = sum(p["funding"] for p in lab_info.get("research_projects", {}).values())
                lab_fundings[lab_name] = lab_fundings.get(lab_name, 0) + funding_sum

max_lab = max(lab_fundings.items(), key=lambda x: x[1])[0]
print(max_lab)

# 27. Print projects grouped by status (ongoing/completed).
grouped_projects = {"ongoing": [], "completed": []}
for uni in university_system["universities"].values():
    for faculty in uni.get("faculties", {}).values():
        for dept in faculty.get("departments", {}).values():
            for lab in dept.get("labs", {}).values():
                for proj_name, proj_info in lab.get("research_projects", {}).items():
                    status = proj_info.get("status")
                    if status in grouped_projects:
                        grouped_projects[status].append(proj_name)

print(grouped_projects)

# 28. Print all courses with more than 100 students enrolled.
popular_courses = []
for uni in university_system["universities"].values():
    for faculty in uni.get("faculties", {}).values():
        for dept in faculty.get("departments", {}).values():
            for course_name, course_info in dept.get("courses", {}).items():
                if course_info.get("students_enrolled", 0) > 100:
                    popular_courses.append(course_name)

print(popular_courses)

# 29. Find total number of faculties in the entire system.
total_faculties = sum(len(uni.get("faculties", {})) for uni in university_system["universities"].values())
print(total_faculties)

# 30. Write a function that searches for a project name and prints its full details.
def search_project(project_name):
    for uni_name, uni in university_system["universities"].items():
        for faculty_name, faculty in uni.get("faculties", {}).items():
            for dept_name, dept in faculty.get("departments", {}).items():
                for lab_name, lab in dept.get("labs", {}).items():
                    projects = lab.get("research_projects", {})
                    if project_name in projects:
                        print(f"Project: {project_name}")
                        print(f"University: {uni_name}")
                        print(f"Faculty: {faculty_name}")
                        print(f"Department: {dept_name}")
                        print(f"Lab: {lab_name}")
                        print(f"Details: {projects[project_name]}")
                        return
    print(f"Project '{project_name}' not found.")

search_project("RoboMind")
```

```python
import pandas as pd
```

```python
df = pd.read_csv("patient_clinical_data_raw.csv")
```

```python
# Columns Containing Missing Values
columnsContaingMissingValues = df.columns[df.isna().any()].tolist()
```

```python
missing_summary = pd.DataFrame({
    'Missing Count': df.isna().sum(),
    'Percentage (%)': df.isna().mean() * 100
})

# Filter out rows with 0 missing values and display
missing_summary[missing_summary['Missing Count'] > 0].sort_values(by='Percentage (%)', ascending=False)
```

```python
df[columnsContaingMissingValues] = df[columnsContaingMissingValues].fillna(df[columnsContaingMissingValues].median())
```

```python
df.columns[df.isna().any()].tolist()
```

```python
# Remove duplicate rows and modify the DataFrame in place
df.drop_duplicates(inplace=True)
```

```python
len(df)
```

```python
# Check the range, min, and max of the Age column
df['Age'].describe()
```

```python
df = df[(df['Age'] >= 1) & (df['Age'] <= 120)]
```

```python
len(df)
```

```python
df['Age'].describe()
```

```python
# Check unique categories and their frequencies for Gender
print("--- Gender Counts ---")
print(df['Gender'].value_counts(dropna=False))

print("\n--- Department Counts ---")
print(df['Department'].value_counts(dropna=False))
```

```python
# Clean the Gender column: remove hidden spaces and force Title Case
df['Gender'] = df['Gender'].astype(str).str.strip().str.title()

# Clean the Department column: remove hidden spaces and force Title Case
df['Department'] = df['Department'].astype(str).str.strip().str.title()
```

```python
print("--- Gender Counts ---")
print(df['Gender'].value_counts(dropna=False))

print("\n--- Department Counts ---")
print(df['Department'].value_counts(dropna=False))
```

```python
# Define the bin edges and the matching labels
bins = [-float('inf'), 17, 40, 60, float('inf')]
labels = ['Child', 'Young Adult', 'Middle Age', 'Senior']

# Create the new column using pd.cut
df['Age_Group'] = pd.cut(df['Age'], bins=bins, labels=labels, right=True)
```

```python
df
```

```python
df['Blood_Pressure']
```

```python
df[['Systolic', 'Diastolic']] = df['Blood_Pressure'].str.split('/', expand=True).astype(int)
```

```python
df.head()
```

```python
import numpy as np 
df['High_Risk'] = np.where(
    (df['Systolic'] >= 140) | 
    (df['Diastolic'] >= 90) | 
    (df['Cholesterol'] >= 240) | 
    (df['Glucose'] >= 126) | 
    (df['Diagnosis'].isin(['Heart Disease', 'Hypertension'])),
    'Yes', 
    'No'
)
```

```python
df.tail()
```

```python
df['Cost_Per_Day']=df['Treatment_Cost']/df['Length_of_Stay']
```

```python
df.tail()
```

```python
print("Average Age: ",df['Age'].mean())
```

```python
print("number of patients in each department", df['Department'].value_counts())
```

```python
# Calculate the average treatment cost grouped by department
df.groupby('Department')['Treatment_Cost'].mean()
```

```python
df.groupby('Diagnosis')['Length_of_Stay'].mean()
```

```python
# Returns the 10 rows with the highest treatment cost
df.nlargest(10, 'Treatment_Cost')
```

```python
# Returns the department name and its average cost
df.groupby('Department')['Treatment_Cost'].mean().nlargest(1)
```

```python
# Calculates the percentage of rows where Readmission equals 'Yes'
readmission_pct = (df['Readmission'] == 'Yes').mean() * 100
print(f"Readmission Rate: {readmission_pct:.2f}%")
```

```python
# Calculate the percentage of readmitted patients for each department
readmission_by_dept = df.groupby('Department').apply(lambda x: (x['Readmission'] == 'Yes').mean() * 100).reset_index(name='Readmission_Rate_(%)')

# Sort from highest readmission rate to lowest
readmission_by_dept.sort_values(by='Readmission_Rate_(%)', ascending=False).round(2)
```

```python
# Calculate the mean cholesterol and glucose levels grouped by diagnosis
df.groupby('Diagnosis')[['Cholesterol', 'Glucose']].mean()
```

```python
# Define clinical thresholds for high levels
high_glucose_cutoff = 126
high_cholesterol_cutoff = 240
high_heart_rate_cutoff = 100

# Filter the DataFrame for patients meeting ALL three conditions
high_risk_patients = df[
    (df['Glucose'] >= high_glucose_cutoff) & 
    (df['Cholesterol'] >= high_cholesterol_cutoff) & 
    (df['Heart_Rate'] > high_heart_rate_cutoff)
]

# Display key details of these patients
high_risk_patients[['Patient_ID', 'Age', 'Diagnosis', 'Glucose', 'Cholesterol', 'Heart_Rate']]
```

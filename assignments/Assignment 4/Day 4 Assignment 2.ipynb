```python
import pandas as pd
import numpy as np

df = pd.read_csv("ecommerce_orders_raw.csv")
```

```python
columnsContaingMissingValues = df.columns[df.isna().any()].tolist()
```

```python
df_duplicate = df[df.duplicated()]
```

```python
df['Payment_Method'].unique()
```

```python
df[columnsContaingMissingValues] = df[columnsContaingMissingValues].fillna(df[columnsContaingMissingValues].median())
```

```python
df['City']=df['City'].astype(str).str.strip().str.title()
```

```python
df['City'].unique()
```

```python
df['Quantity'].describe()
```

```python
df= df[df['Quantity']>0]
```

```python
df['Quantity'].describe()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.shape
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
```

```python
df['Gross_Amount'] = df['Quantity']*df['Unit_Price']
```

```python
df.head()
```

```python
df['Discount_Amount'] = (df['Gross_Amount']*df['Discount'])/100
```

```python
df.head()
```

```python
df['Net_Amount'] = df['Gross_Amount']-df['Discount_Amount']
```

```python
df.head()
```

```python
bins = [-float('inf'), 2, 3, 4, float('inf')]
labels = ['Poor', 'Average', 'Good', 'Excellent']
```

```python
df['Rating_Category'] = pd.cut(df['Rating'], bins=bins, labels=labels, right=True)
```

```python
df.head()
```

```python
# 1. Extract the four-digit year (e.g., 2025, 2026)
df['Year'] = df['Order_Date'].dt.year

# 2. Extract the month as a number (1 to 12)
df['Month'] = df['Order_Date'].dt.month

# 3. Extract the day of the month (1 to 31)
df['Day'] = df['Order_Date'].dt.day

# 4. Extract the name of the day of the week (e.g., Thursday, Saturday)
df['Day_of_Week'] = df['Order_Date'].dt.day_name()

# Preview the new columns
print(df[['Order_Date', 'Year', 'Month', 'Day', 'Day_of_Week']].head())
```

```python
total_revenue = df['Net_Amount'].sum()
print(f"Total Revenue: Rs. {total_revenue:,.2f}")
```

```python
# Uses nunique() in case an order spans multiple rows/items
total_orders = df['Order_ID'].nunique() 
print(f"Total Orders: {total_orders}")
```

```python
# Total revenue divided by the unique number of orders
average_order_value = df['Net_Amount'].mean()
print(f"Average Order Value: Rs. {average_order_value:,.2f}")
```

```python
revenue_by_category = df.groupby('Product_Category')['Net_Amount'].sum().sort_values(ascending=False)
print(revenue_by_category)
```

```python
top_10_products = df.groupby('Product')['Net_Amount'].sum().nlargest(10)
print(top_10_products)
```

```python
# Groups by city, sums revenue, and grabs the top index name
highest_revenue_city = df.groupby('City')['Net_Amount'].sum().idxmax()
print(f"Highest Revenue City: {highest_revenue_city}")
```

```python
# Grabs the single most common payment method
top_payment_method = df['Payment_Method'].value_counts().idxmax()
print(f"Most Frequent Payment Method: {top_payment_method}")
```

```python
avg_rating_by_category = df.groupby('Product_Category')['Rating'].mean().sort_values(ascending=False)
print(avg_rating_by_category)
```

```python
# Grouping by both Year and Month ensures historical trends stay separated chronologically
monthly_revenue = df.groupby(['Year', 'Month'])['Net_Amount'].sum().reset_index()
print(monthly_revenue)
```

```python
import calendar
highest_month = df.groupby(['Year', 'Month'])['Net_Amount'].sum().idxmax()
year, month = highest_month
print(f"Highest Revenue Month (Year, Month): {calendar.month_name[month]} {year}")
```

```python
top_10_customers = df.groupby('Customer_ID')['Net_Amount'].sum().nlargest(10)
print(top_10_customers)
```

```python
# Returns a sorted comparison list of all cities and their financial contributions
city_comparison = df.groupby('City')['Net_Amount'].sum().sort_values(ascending=False).reset_index()
print(city_comparison)
```

```python
# First calculate AOV per category: Total Category Revenue / Unique Orders in that category
category_aov = df.groupby('Product_Category').apply(lambda x: x['Net_Amount'].sum() / x['Order_ID'].nunique())
highest_category_aov = category_aov.idxmax()

print(f"Category with Highest AOV: {highest_category_aov} Rs. {category_aov.max():,.2f}")
```

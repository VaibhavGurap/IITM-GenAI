## Task 3 — Create a 3D Tensor
Now suppose we monitor 10 machines. Each machine has:
* 5 time steps

Create a tensor with shape `(10, 5, 4)` interpreting dimensions as: **Machine $\times$ Time $\times$ Sensor**.

## Task 4 — Indexing a Tensor
Create:
`X = np.arange(24).reshape(2, 3, 4)`

Print `X`, then investigate `X[0]`, `X[1]`, `X[0, 0]`, and `X[0, 0, 0]`.

### Answers:
* **Shape of X:** `(2, 3, 4)`
* **`X[0]`:** The first 2D block/matrix of shape `(3, 4)`.

## Task 8 — Reshape
Create `X = np.arange(24)`. Reshape it into `(4, 6)`, `(2, 3, 4)`, `(6, 4)`, and `(24,)`.

Print tensor, shape, and ndim for each.

### Question:
**Why does reshaping change the shape but not the total number of elements?**
> *Explanation:* Reshaping simply alters the indexing scheme / memory layout metadata (how dimensions are viewed) while preserving the underlying contiguous 1D array of elements in memory. As long as the product of the new dimensions equals the total element count ($4 \times 6 = 2 \times 3 \times 4 = 24$), the operation is valid.

```python
X = np.arange(24)

shapes = [(4, 6), (2, 3, 4), (6, 4), (24,)]

for s in shapes:
    reshaped = X.reshape(s)
    print(f"--- Reshaped to {s} ---")
    print("Shape:", reshaped.shape)
    print("NDim:", reshaped.ndim)
    print("Total Elements:", reshaped.size)
    print("Values:
", reshaped, "
")
```

## Task 9 — Reshape vs Transpose
Create `X = np.arange(24).reshape(2, 3, 4)`.
Perform `X.reshape(3, 2, 4)` and `X.transpose(1, 0, 2)`. Compare results.

### Are these two operations doing the same thing?
> **No!**
> * `reshape(3, 2, 4)` re-interprets the raw memory buffer linearly into new shape dimensions.
> * `transpose(1, 0, 2)` physically permutes/swaps the axes (axis 0 becoming axis 1 and vice versa), preserving spatial structural relationships across axes.

```python
X = np.arange(24).reshape(2, 3, 4)

reshaped_X = X.reshape(3, 2, 4)
transposed_X = X.transpose(1, 0, 2)

print("Original X[0, 1, 2]:", X[0, 1, 2])
print("Reshaped X shape:", reshaped_X.shape)
print("Transposed X shape:", transposed_X.shape)
print("
Are values identical?", np.array_equal(reshaped_X, transposed_X))
print("
Reshaped X:
", reshaped_X)
print("
Transposed X:
", transposed_X)
```

## Task 10 — Broadcasting
Create `X = np.ones((3, 4))` and `b = np.array([10, 20, 30, 40])`.
Calculate `X + b` and `X * b`.

### Question: Why can a (4,) tensor be combined with a (3, 4) tensor?
> *Explanation:* According to NumPy broadcasting rules, trailing dimensions are compared right-to-left. The 1D tensor of shape `(4,)` matches the last dimension of `(3, 4)`. NumPy automatically virtual-stretches/replicates `b` across axis 0 to match shape `(3, 4)` without allocating extra memory.

```python
X = np.ones((3, 4))
b = np.array([10, 20, 30, 40])

add_result = X + b
mult_result = X * b

print("X + b:
", add_result)
print("
X * b:
", mult_result)
```

## Task 12 — Tensor Aggregation
Create `X = np.random.randn(100, 20, 5)` (100 machines, 20 time steps, 5 sensors).

Calculate:
* Mean across time (axis 1) $\rightarrow$ expected shape: `(100, 5)`
* Mean across sensors (axis 2) $\rightarrow$ expected shape: `(100, 20)`
* Mean across machines (axis 0) $\rightarrow$ expected shape: `(20, 5)`
* Maximum sensor value across time (axis 1) $\rightarrow$ expected shape: `(100, 5)`

## Task 13 — Normalize a Tensor
Create sensor data `X = np.random.randn(100, 20, 5)`.
Calculate mean and standard deviation for each sensor (across machines and time, i.e., axes (0, 1)).

Normalize using:
$$X_{\text{norm}} = \frac{X - \mu}{\sigma}$$

Verify $\text{mean} \approx 0$ and $\text{std} \approx 1$.

```python
X_norm_data = np.random.randn(100, 20, 5) * 5.0 + 10.0  # arbitrary mean=10, std=5

# Calculate mean & std per sensor across axes 0 and 1
mean_per_sensor = X_norm_data.mean(axis=(0, 1), keepdims=True)
std_per_sensor = X_norm_data.std(axis=(0, 1), keepdims=True)

# Standardize
X_normalized = (X_norm_data - mean_per_sensor) / std_per_sensor

print("Normalized mean per sensor:", X_normalized.mean(axis=(0, 1)))
print("Normalized std per sensor:", X_normalized.std(axis=(0, 1)))
```

## Task 14 — Flattening
Create an image `(28, 28, 3)` and flatten it into `(2352,)`.
Then create a batch of 64 images `(64, 28, 28, 3)` and flatten each image while keeping batch dimension $\rightarrow$ `(64, 2352)`.

### Why shouldn't we simply flatten the entire tensor into (150528,) when working with a batch?
> *Explanation:* Flattening the entire batch destroys the individual sample boundary distinction. Machine learning batch processing requires each sample in the batch to remain an independent row/vector so matrix multiplication and activation calculations act on each item individually.

```python
img_single = np.random.randn(28, 28, 3)
img_single_flat = img_single.flatten()
print("Single image flattened shape:", img_single_flat.shape)

batch_img = np.random.randn(64, 28, 28, 3)
# Reshape preserving batch dimension (axis 0)
batch_img_flat = batch_img.reshape(batch_img.shape[0], -1)

print("Batch flattened shape (preserving batch dim):", batch_img_flat.shape)
```

## Task 15 — Concatenation
Combine visual features `(32, 128)` and sensor features `(32, 64)` into `(32, 192)` along feature dimension.

### Why can we concatenate along the feature dimension but not arbitrarily along every dimension?
> *Explanation:* Concatenation requires all dimensions *except* the target concatenation axis to match exactly. Here, both tensors have batch size 32 (axis 0 matches), so we can concatenate along axis 1 (features). We could not concatenate along axis 0 because the feature dimensions (128 vs 64) do not match.

```python
visual_features = np.random.randn(32, 128)
sensor_features = np.random.randn(32, 64)

combined_features = np.concatenate([visual_features, sensor_features], axis=1)

print("Visual Features Shape:", visual_features.shape)
print("Sensor Features Shape:", sensor_features.shape)
print("Combined Features Shape:", combined_features.shape)
```

## Task 16 — Practical Mini Challenge: Sensor Dataset
Create synthetic dataset `X.shape = (500, 60, 8)` representing 500 machines, 60 time steps, 8 sensors.

Perform:
1. Extract machine 10.
2. Extract the first 20 time steps.
3. Extract sensor 5 from all machines.
4. Calculate average value of every sensor.
5. Calculate average value at every time step.
6. Find machine with highest average sensor reading.
7. Normalize each sensor.
8. Split tensor into Training (400 machines) and Testing (100 machines).

```python
np.random.seed(42)
dataset = np.random.randn(500, 60, 8) * 2.0 + 50.0

# 1. Machine 10 (idx 9)
m10 = dataset[9]
print("1. Machine 10 shape:", m10.shape)

# 2. First 20 time steps
t20 = dataset[:, :20, :]
print("2. First 20 time steps shape:", t20.shape)

# 3. Sensor 5 (idx 4) from all machines
s5 = dataset[:, :, 4]
print("3. Sensor 5 shape:", s5.shape)

# 4. Average value of every sensor (across machines and time)
avg_sensors = dataset.mean(axis=(0, 1))
print("4. Average of each sensor:
", avg_sensors)

# 5. Average value at every time step (across machines and sensors)
avg_timesteps = dataset.mean(axis=(0, 2))
print("5. Average value at each time step shape:", avg_timesteps.shape)

# 6. Machine with highest average sensor reading
machine_means = dataset.mean(axis=(1, 2))
top_machine_idx = np.argmax(machine_means)
print(f"6. Machine with highest average reading: Machine index {top_machine_idx} (Mean: {machine_means[top_machine_idx]:.4f})")

# 7. Normalize each sensor
mean_s = dataset.mean(axis=(0, 1), keepdims=True)
std_s = dataset.std(axis=(0, 1), keepdims=True)
dataset_norm = (dataset - mean_s) / std_s
print("7. Sensor normalization verified mean:", np.round(dataset_norm.mean(axis=(0, 1)), 4))

# 8. Train / Test Split
train_set = dataset_norm[:400]
test_set = dataset_norm[400:]
print("8. Train set shape:", train_set.shape, "| Test set shape:", test_set.shape)
```

### What could `X = torch.randn(32, 100, 768)` represent in Deep Learning?

In Natural Language Processing (NLP) or Transformer architectures (such as BERT or GPT), these three dimensions typically represent:

1. **`32` (Batch Size):** The number of sequences/sentences processed simultaneously in a single forward pass.
2. **`100` (Sequence Length / Tokens):** The number of words/tokens per sequence (e.g., 100 words/tokens per sentence).
3. **`768` (Embedding Dimension / Hidden Size):** The vector representation size for each token (e.g., standard BERT-base hidden dimensionality).
